```python
# Veri setini yükleme
data = pd.read_csv('dataset/BTK-Datathon-2024/train.csv',low_memory=False)
```

```python
import pandas as pd
import numpy as np


# 1. Veri setini inceleme
print(data.info())
print(data.describe())
print(data.isnull().sum())

# 2. Eksik değerlerin işlenmesi
# Eksik değer oranını hesaplama
missing_ratio = data.isnull().sum() / len(data) * 100
print(missing_ratio)

# Eksik değerleri doldurma (örnek: sayısal sütunlar için medyan, kategorik sütunlar için mod)
for col in data.select_dtypes(include=['float64', 'int64']).columns:
    data[col].fillna(data[col].median(), inplace=True)

for col in data.select_dtypes(include=['object']).columns:
    data[col].fillna(data[col].mode()[0], inplace=True)

# 3. Kategorik değişkenlerin işlenmesi
# One-hot encoding
data = pd.get_dummies(data, drop_first=True)

# 4. Sayısal değişkenlerin işlenmesi
# Aykırı değerleri tespit etme ve işleme
for col in data.select_dtypes(include=['float64', 'int64']).columns:
    q1 = data[col].quantile(0.25)
    q3 = data[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    data[col] = np.where(data[col] < lower_bound, lower_bound, data[col])
    data[col] = np.where(data[col] > upper_bound, upper_bound, data[col])

# 5. Yeni özellikler oluşturma
# Örnek: Yaş hesaplama
data['Dogum Tarihi'] = pd.to_datetime(data['Dogum Tarihi'], errors='coerce')
data['Yas'] = 2024 - data['Dogum Tarihi'].dt.year

# 6. Gereksiz sütunları çıkarma
data.drop(['Dogum Tarihi', 'id'], axis=1, inplace=True)

# Temizlenmiş veri setini kaydetme
data.to_csv('cleaned_dataset.csv', index=False)

print("Veri seti temizleme işlemi tamamlandı.")
```
